```python
from pyspark.sql.functions import *
```

```python
raw_df = spark.read.table("workspace.default.online_retail")
```

```python
raw_df.limit(10).display()
```

```python
raw_df.printSchema()
```

```python
raw_df.describe().show()
```

```python
df_silver = raw_df.filter("CustomerID IS NOT NULL")
```

```python
df_silver = raw_df.filter("Quan")
```

```python
df_silver = raw_df.filter('UnitPrice > 0')
```

```python
df_silver = raw_df.withColumn(
    'Total_Amount',
    col('Quantity') * col('UnitPrice')
)
```

```python
df_silver.limit(10).display()
```

```python
df_silver.describe().show()
```

```python
df_silver.write \
    .mode("overwrite") \
    .format("delta") \
    .saveAsTable("workspace.default.online_retail_silver")
```

```python
spark.table("workspace.default.online_retail_silver").show(5)
```

```python
df_silver.limit(10).display()
```

```python
spark.sql("DESCRIBE DETAIL workspace.default.online_retail_silver").show(truncate=False)
```

```python
df_silver.show(5)
```

```python
daily_revenue = (
    df_silver.withColumn("order_date", to_date("InvoiceDate"))
      .groupBy("order_date")
      .agg(sum("total_amount").alias("daily_revenue"))
)
```

```python
daily_revenue.write \
    .mode("overwrite") \
    .format("delta") \
    .saveAsTable("workspace.default.gold_daily_revenue")
```

```python
country_revenue = (
    df_silver.groupBy("Country")
      .agg(sum("total_amount").alias("country_revenue"))
      .orderBy("country_revenue", ascending=False)
)
```

```python
country_revenue.write \
    .mode("overwrite") \
    .format("delta") \
    .saveAsTable("workspace.default.gold_country_revenue")
```

```python
product_revenue = (
    df_silver.groupBy("StockCode", "Description")
      .agg(sum("total_amount").alias("product_revenue"))
      .orderBy("product_revenue", ascending=False)
)
```

```python
product_revenue.write \
    .mode("overwrite") \
    .format("delta") \
    .saveAsTable("workspace.default.gold_product_revenue")
```

```python
spark.table("workspace.default.gold_daily_revenue").show(5)
spark.table("workspace.default.gold_country_revenue").show(5)
spark.table("workspace.default.gold_product_revenue").show(5)
```
